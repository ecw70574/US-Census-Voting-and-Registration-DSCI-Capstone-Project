```python
import pandas as pd
import numpy as np
from fredapi import Fred

fred = Fred(api_key=YOUR_API_KEY)

states = [
    "AL","AK","AZ","AR","CA","CO","CT","DE","FL","GA",
    "HI","ID","IL","IN","IA","KS","KY","LA","ME","MD",
    "MA","MI","MN","MS","MO","MT","NE","NV","NH","NJ",
    "NM","NY","NC","ND","OH","OK","OR","PA","RI","SC",
    "SD","TN","TX","UT","VT","VA","WA","WV","WI","WY"
]
```

```python
def get_fred_series(series_id, state):
    s = fred.get_series(series_id)
    df = s.reset_index()
    df.columns = ["date", "value"]
    df["state"] = state
    df["value"] = pd.to_numeric(df["value"], errors="coerce")
    return df
```

```python
unemp_list = []
for s in states:
    try:
        df = get_fred_series(f"{s}UR", s)
        unemp_list.append(df)
    except:
        pass

unemp = pd.concat(unemp_list)
```

```python
unemp
```

```python
income_list = []
for s in states:
    try:
        df = get_fred_series(f"{s}PCPI", s)
        income_list.append(df)
    except:
        pass

income = pd.concat(income_list)
```

```python
income["date"] = pd.to_datetime(income["date"])
unemp["date"] = pd.to_datetime(unemp["date"])
```

```python
income["year"] = income["date"].dt.year
unemp["year"] = unemp["date"].dt.year

unemp["month"] = unemp["date"].dt.month
```

```python
income
```

```python
unemp
```

```python
income_filteryear = income[(income["year"] >= 2010) & (income["year"] <= 2024)]
income_filteryear
```

```python
unemp_filteryear = unemp[(unemp["year"] >= 2010) & (unemp["year"] <= 2024)]
unemp_filteryear
```

```python
unemp_octnov = unemp_filteryear[(unemp_filteryear["month"] == 10) | (unemp_filteryear["month"] == 11)]
unemp_octnov
```

```python
#rename months 
unemp_octnov.loc[unemp_octnov['month'] == 10, 'newmonth'] = 'october'
unemp_octnov.loc[unemp_octnov['month'] == 11, 'newmonth'] = 'november'

# remove date columns 
unemp_octnov = unemp_octnov.drop(columns=["date", "month"])
unemp_octnov = unemp_octnov.rename(columns={"newmonth":"month", "value":"unemployment_pct"})
income_filteryear = income_filteryear.reset_index()
income_filteryear["value"] = income_filteryear["value"].astype(int)
income_filteryear = income_filteryear[["value", "state", "year"]]

income_filteryear = income_filteryear.rename(columns={"value":"percapita_personalincome"})
```

```python
income_filteryear
```

```python
# flatten month
wide = unemp_octnov.pivot_table(
    index=["state", "year"],
    columns="month",
    values="unemployment_pct"
).reset_index()
wide = wide.rename(columns={"november":"nov_unemp", "october":"oct_unemp"})
```

```python
wide
```

```python
# merge on year and state
merged = pd.merge(income_filteryear, wide, on=["state","year"])
merged
```

```python
# only keep even years for join with elections/midterm years
merged = merged[(merged["year"] == 2010) | (merged["year"] == 2012) | (merged["year"] == 2014) | (merged["year"] == 2018) | (merged["year"] == 2020) | (merged["year"] == 2022) |(merged["year"] == 2024)]
# make year type consistent for merge 
merged.loc[merged['year'] == 2010, 'year'] = 10
merged.loc[merged['year'] == 2012, 'year'] = 12
merged.loc[merged['year'] == 2014, 'year'] = 14
merged.loc[merged['year'] == 2016, 'year'] = 16
merged.loc[merged['year'] == 2018, 'year'] = 18
merged.loc[merged['year'] == 2020, 'year'] = 20
merged.loc[merged['year'] == 2022, 'year'] = 22
merged.loc[merged['year'] == 2024, 'year'] = 24

```

```python
merged = merged[["state","year","oct_unemp","nov_unemp","percapita_personalincome"]]
merged.head(10)
```
